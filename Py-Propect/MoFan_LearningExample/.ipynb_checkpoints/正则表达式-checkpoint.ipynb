```python
import re
```

```python
pattern1 = "cat"
pattern2 = "bird"
string = "dog runs to cat"
print(pattern1 in string)
print(pattern2 in string)
```

```python
pattern1 = "cat"
pattern2 = "bird"
string = "dog runs to cat"
print(re.search(pattern1, string))#使用正则表达式,找到了就返回对象，没找到就返回none
print(re.search(pattern2, string))
```

```python
ptn = r"r[au]n"#最前面加了一个r表示是正则表达式其中[au]表示run和ran两种形式
print(re.search(ptn, "dog rans to cat"))#
```

```python
#同上[8]类似
print(re.search(r"r[A-Z]n", "dog runs to cat"))
print(re.search(r"r[0-9a-z]n", "dog runs to cat"))
```

```python
#\d表示匹配所有0到9的所有数字，也就是和r"r[0-9]n"是一样的
print(re.search(r"r\dn", "run r9n"))
#\D表示匹配除数字以外的所有符号，如英文字母
print(re.search(r"r\Dn", "run r9n"))
```

```python
#\s匹配所有空白符，例如换行符，tab符号等
print(re.search(r"r\sn", "r\nn r9n"))   
#\S匹配所有不是空白符的东西
print(re.search(r"r\Sn", "r\nn run"))
```

```python
#\w是匹配[a-z],[A-Z],[0-9]和下横杠线_
#\W表示除了以上东西都可以匹配
```

```python
#\b没看懂
print(re.search(r"\b runs \b", "dog runs to cat"))
print(re.search(r"\b runs\b", "dog runs to cat"))
#B
print(re.search(r"\B runs \B", "dog  runs  to cat"))
```
